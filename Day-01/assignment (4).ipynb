```python
df['diagnosis'] = df['diagnosis'].map({'M': 1, 'B': 0})
df['diagnosis'].value_counts() # As the diagnosis is a categorical variable trying to preform encoding
```

```python
sns.boxplot(df["perimeter_mean"]) # we can see it has some outliers
```

```python
q1=df["perimeter_mean"].quantile(0.25)
q3=df["perimeter_mean"].quantile(0.75)
iqr=q3-q1
upper_bound=q3+1.5*iqr
lower_bound=q1-1.5*iqr
df["perimeter_mean"]=df["perimeter_mean"].clip(upper=upper_bound,lower=lower_bound)
```

```python
sns.boxplot(df["perimeter_mean"]) # we can see that outliers are now removed
```

```python
df.corr() # we can check the corelations between each pair of attricutes
```

```python
df['radius_ratio'] = df['radius_worst'] / df['radius_mean']

# Feature 2: Difference between worst area and mean area
df['area_difference'] = df['area_worst'] - df['area_mean']

# Display newly engineered features
df[['radius_ratio', 'area_difference']].head()
```

1. Dataset Size: The dataset contains 569 rows and 32 columns, including patient/sample ID, diagnosis, and 30 numerical features.

2. No Missing Values: All 32 columns have 569 non-null values, indicating there are no missing values.

3. Target Variable: diagnosis is the categorical target variable, typically representing Malignant (M) or Benign (B) tumors.

4. Feature Categories: The 30 numerical features are divided into three groups: mean, standard error (se), and worst, with 10 measurements in each group.

5. Data Preprocessing: The id column can be dropped because it is an identifier. The diagnosis column should be encoded into numerical labels, and feature scaling is recommended for distance-based or scale-sensitive ML models.
