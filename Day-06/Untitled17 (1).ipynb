```python
df=df.dropna()
```

```python
stop_words = set(stopwords.words('english'))
stop_words = stop_words - {'no', 'nor', 'not'}
stemmer = PorterStemmer()

def preprocess_text(text):
    text = str(text).lower()
    text = re.sub(r'<.*?>', ' ', text)             # Remove HTML tags
    text = re.sub(r'https?://\\S+|www\\.\\S+', ' ', text)  # Remove URLs
    text = re.sub(r'\\d+', ' ', text)             # Remove digits
    text = text.translate(str.maketrans('', '', string.punctuation))

    tokens = nltk.word_tokenize(text)

    cleaned_tokens = [
        stemmer.stem(word)
        for word in tokens
        if word.isalpha() and word not in stop_words
    ]

    return ' '.join(cleaned_tokens)

# Preview preprocessing on one review
print("Original review:\\n", df['Review Text'].iloc[0][:700])
print("\\nCleaned review:\\n", preprocess_text(df['Review Text'].iloc[0])[:700])
```

```python
df['Review Text'] = df['Review Text'].apply(preprocess_text)
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
```

```python
def remove(txt):
  words = txt.split()
  cleaned = []
  for i in words:
    if not i in stop_words:
      cleaned.append(i)

  return ' '.join(cleaned)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df['Review Text'], df['Rating'], test_size=0.20, random_state=42)
```

```python
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
bow=CountVectorizer()
X_train_bow=bow.fit_transform(X_train)
X_test_bow=bow.transform(X_test)
```

```python
print(X_train_bow)
```

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression(max_iter=1000)
lr.fit(X_train_bow,y_train)
lr.score(X_test_bow,y_test)
```

```python
tfidf=TfidfVectorizer()
X_train_tf=tfidf.fit_transform(X_train)
X_test_tf=tfidf.transform(X_test)
```

```python
lr1=LogisticRegression()
lr1.fit(X_train_tf,y_train)
lr1.score(X_test_tf,y_test)
```

```python
y_pred=lr1.predict(X_test_tf)
```

```python
from sklearn.metrics import classification_report, f1_score
print("F1 Score:", f1_score(y_test, y_pred, average='weighted'))
print(classification_report(y_test, y_pred))
```

```python
#So we can see that compared to normal count vectorization tfidf gave better results
```

```python
# Top 10 important words are
feature_names = tfidf.get_feature_names_out()

# Get Logistic Regression coefficients
coefficients = lr1.coef_[0]

# Top 10 positive sentiment words
top_positive = np.argsort(coefficients)[-10:][::-1]

# Top 10 negative sentiment words
top_negative = np.argsort(coefficients)[:10]

# Create comparison table
top_words = pd.DataFrame({
    'Positive Words': feature_names[top_positive],
    'Positive Weight': coefficients[top_positive],
    'Negative Words': feature_names[top_negative],
    'Negative Weight': coefficients[top_negative]
})

print(top_words)
```
