```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['Extracurricular Activities']=le.fit_transform(df['Extracurricular Activities'])
```

```python
df.head(5)
```

```python
x=df.drop(columns=['Performance Index'])
y=df['Performance Index']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.20, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train, y_train)
lr_pred=lr.predict(x_test)
lr_pred
```

```python
from sklearn.linear_model import Ridge, Lasso, ElasticNet
rd=Ridge(alpha=1.0)
ls=Lasso(alpha=1.0)
en=ElasticNet(alpha=1.0, l1_ratio=0.5)
rd.fit(x_train, y_train)
rd_pred=rd.predict(x_test)
rd_pred
```

```python
ls.fit(x_train, y_train)
ls_pred=ls.predict(x_test)
ls_pred
```

```python
en.fit(x_train, y_train)
en_pred=en.predict(x_test)
en_pred
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

lr_mae = mean_absolute_error(y_test, lr_pred)
lr_mse = mean_squared_error(y_test, lr_pred)
lr_rmse = np.sqrt(lr_mse)
lr_r2 = r2_score(y_test, lr_pred)

rd_mae = mean_absolute_error(y_test, rd_pred)
rd_mse = mean_squared_error(y_test, rd_pred)
rd_rmse = np.sqrt(rd_mse)
rd_r2 = r2_score(y_test, rd_pred)

ls_mae = mean_absolute_error(y_test, ls_pred)
ls_mse = mean_squared_error(y_test, ls_pred)
ls_rmse = np.sqrt(ls_mse)
ls_r2 = r2_score(y_test, ls_pred)

en_mae = mean_absolute_error(y_test, en_pred)
en_mse = mean_squared_error(y_test, en_pred)
en_rmse = np.sqrt(en_mse)
en_r2 = r2_score(y_test, en_pred)
```

```python
comparison = pd.DataFrame({
    'Model': ['Linear Regression', 'Ridge', 'Lasso', 'Elastic Net'],
    'MAE': [lr_mae, rd_mae, ls_mae, en_mae],
    'MSE': [lr_mse, rd_mse, ls_mse, en_mse],
    'RMSE': [lr_rmse, rd_rmse, ls_rmse, en_rmse],
    'R²': [lr_r2, rd_r2, ls_r2, en_r2]
})

print(comparison.round(4).to_string(index=False))
```

Linear Regression and Ridge Regression achieved the best performance, with the lowest RMSE (2.0206) and highest R² (0.9890). Among the regularization models: Ridge regularization had no noticeable effect on performance, while Lasso and Elastic Net slightly increased prediction errors and reduced R². Overall, Linear Regression is the preferred model due to its simplicity and equally strong performance.
